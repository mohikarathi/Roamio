# 05. Recommendation Error Analysis & System Limitations

### Objective
Conduct an honest post-mortem analysis of recommendation failures, identifying where individual components fail and documenting architectural trade-offs.

## 1. Where TF-IDF Outperforms Embeddings
- **Exact Proper Noun Entities & Acronyms**: If a user queries a specific landmark (e.g. "Kinkaku-ji" or "Angkor Wat"), TF-IDF provides an exact lexical match.
- **High-Frequency Stopwords in Dense Queries**: Dense models occasionally suffer from semantic drift when complex multi-clause constraints are concatenated.

## 2. Where Dense Retrieval Fails
- **Budget Blindness**: A user asking for "luxury beachfront overwater villas" will match the Maldives, even if their stated budget is ₹20,000. Dense retrieval has no intrinsic concept of money.
- **Geographical Leakage**: Queries like "warm beaches in Asia" can trigger "Cancun, Mexico" or "Copacabana, Brazil" if the textual similarity overrides weak geography tokens without two-stage filtering.

## 3. Limitations & Future Scope
- **Live Flight Pricing**: Real flight ticket prices vary daily; Roamio currently relies on estimated daily ground costs in INR.
- **Dynamic Weather Anomalies**: Uses historic best travel months rather than live satellite radar weather APIs.
- **Learning to Rank (LTR)**: With real user clickstream and conversion data, the linear weights (..w_7$) can be trained using LambdaMART or RankNet.
