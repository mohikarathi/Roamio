```python
from src.retrieval.tfidf import TFIDFRetriever
from src.retrieval.dense import DenseRetriever

tfidf = TFIDFRetriever()
dense = DenseRetriever()

test_query = "serene alpine retreat with mountain trekking"
print(f"Query: {test_query}
")

print("--- TF-IDF Top 3 ---")
for r in tfidf.retrieve(test_query, top_k=3):
    print(f"{r.destination.name} ({r.destination.country}): {r.score:.4f}")

print("
--- Dense Embeddings Top 3 ---")
for r in dense.retrieve(test_query, top_k=3):
    print(f"{r.destination.name} ({r.destination.country}): {r.score:.4f}")
```

## 2. Analysis of Results

- Notice that the query uses words like **"serene"** and **"alpine retreat"**.
- TF-IDF fails when documents use **"peaceful"**, **"Matterhorn"**, or **"Swiss Alps"** if the exact words do not overlap.
- The Dense Sentence Transformer captures the semantic geometry: "serene" $pprox$ "peaceful", "alpine retreat" $pprox$ "mountain resort".
